```python
import os
from pathlib import Path

REQUIRED_ENV_VARS = []  # 例如：["API_KEY", "DATABASE_URL"]

try:
    from dotenv import dotenv_values, load_dotenv
except ImportError:
    print("未安装 python-dotenv。请先运行：%pip install python-dotenv")
else:
    env_file = next(
        (parent / ".env" for parent in [Path.cwd(), *Path.cwd().parents]
         if (parent / ".env").is_file()),
        None,
    )

    if env_file:
        dotenv_keys = [
            key for key, value in dotenv_values(env_file).items()
            if key and value is not None
        ]
        load_dotenv(env_file, override=False)
        print(f"找到 .env：{env_file}")
        print(f".env 中已配置 {len(dotenv_keys)} 个变量。")
    else:
        dotenv_keys = []
        print("未找到 .env 文件；将仅检查当前进程环境变量。")

    missing = [
        name for name in set(dotenv_keys + REQUIRED_ENV_VARS)
        if not os.getenv(name)
    ]

    if missing:
        print("以下变量未加载或为空：", ", ".join(sorted(missing)))
    else:
        print("环境变量检查通过。")
```
